# Matplotlib

```python
import matplotlib.pyplot as plt
```

### The Basic Line plot:

```python
x = [1, 2, 3, 4, 5]
y = [10, 20, 25, 30, 40]
plt.plot(x, y)  # Line plot
plt.show()
```

```python
x = [1, 2, 3, 4, 5]
y = [10, 20, 25, 30, 40]
plt.plot(x,y)
plt.title("Basic Line Plot")
plt.xlabel("X-axis")
plt.ylabel("Y-axis")
plt.grid(False) # الخطوط الخلفية 
plt.show()
```

### Multible Lines :

```python
x = [1, 2, 3, 4, 5]
y1 = [1, 2, 3, 4, 5]
y2 = [1, 4, 9, 16, 25]
plt.plot(x, y1, label='Linear' )
plt.plot(x, y2, label='Squares')
plt.legend()
plt.show()
```

```python
x = [1, 2, 3, 4, 5]
y1 = [1, 2, 3, 4, 5]
y2 = [1, 4, 9, 16, 25]
y3 = [1, 8, 27, 64, 125]
plt.plot(x, y1, label='Linear' , linestyle='-.',color='yellow')
plt.plot(x, y2, label='Squares',linestyle=':' , color='b') 
plt.plot(x, y3, label='cubes' , linestyle='--', color="#8D38C9",lw=3)
plt.legend()
plt.show()
```

# The Type of Bars : 
## 1-Bar Chart 

```python
categories = ['A', 'B', 'C']
values = [5, 7, 3]

plt.bar(categories, values)
plt.title("Bar Chart")
plt.show()
```

### Bar Chart (Horizantal) :

```python
plt.barh(x, y, color='orange')
plt.title("Horizontal Bar Chart")
plt.show()
```

# 2- Histogram

```python
data = [1, 2, 2, 3, 3, 3, 4, 4, 5]

plt.hist(data, bins=5, edgecolor='black')
plt.title("Histogram")
plt.show()
```

# 3-Pi Chart :

```python
sizes = [30, 25, 20, 25]
plt.figure(figsize=(5,5))
plt.pie(sizes)
plt.title("Pie Chart")
plt.show()
```

```python
sizes = [30, 25, 20, 25]
labels = ["Apples", "Bananas", "Cherries", "Dates"]
plt.figure(figsize=(5,5))
plt.pie(sizes, labels=labels)
plt.title("Pie Chart")
plt.show()
```

```python
sizes = [30, 25, 20, 25]
labels = ["Apples", "Bananas", "Cherries", "Dates"]
plt.figure(figsize=(5,5))
plt.pie(sizes, labels=labels,autopct='%1.1f%%')
plt.title("Pie Chart")
plt.show()
```

```python
sizes = [40, 25, 20, 15]
labels = ["Math", "Physics", "Chemistry", "Biology"]
colors = ["gold", "lightblue", "lightgreen", "pink"]
plt.pie(sizes, labels=labels, colors=colors, autopct="%1.1f%%")
plt.title("Student Subject Distribution")
plt.show()
```

```python
sizes = [40, 25, 20, 15]
labels = ["Math", "Physics", "Chemistry", "Biology"]
colors = ["gold", "lightblue", "lightgreen", "pink"]
explode = [0,0,0.2,0]
plt.pie(sizes, labels=labels, colors=colors, autopct="%1.1f%%", explode=explode,shadow=True,startangle=90)
plt.title("Student Subject Distribution")
plt.savefig("pie_without_figsize.png")
plt.show()
```

```python
import matplotlib.pyplot as plt

# البيانات
sizes = [30, 20, 25, 25]
labels = ["A", "B", "C", "D"]

# رسم Donut chart
plt.figure(figsize=(5,5))
plt.pie(sizes, labels=labels, wedgeprops={"width":0.4})  # width =سمك الشرائح 

plt.title("Donut Chart")
plt.show()
```

# Scatter 

```python
import numpy as np

x = np.random.rand(100)
y = np.random.rand(100)

plt.scatter(x, y, alpha=0.5) 
plt.show()
```

```python
# بيانات عشوائية
np.random.seed(0)
x = np.random.rand(20) * 10   
y = np.random.rand(20) * 10   
sizes = np.random.rand(20) * 1000 
colors = np.random.rand(20)       

plt.scatter(x, y, s=sizes, c=colors, alpha=0.5,  edgecolors="k")

plt.xlabel("Revenue")
plt.ylabel("Profit")
plt.title("Bubble Chart")
plt.grid(True)
plt.show()
```
